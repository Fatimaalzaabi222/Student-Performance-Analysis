```python
def Desc_stat(ds, var):
    mean = ds[var].mean()
    median = ds[var].median()
    mode = ds[var].mode()
    minimum = ds[var].min()
    maximum = ds[var].max()
    range = maximum - minimum
    std_dev = ds[var].std()
    variance = ds[var].var()
    skew = ds[var].skew()
    kurtosis = ds[var].kurt()

    Quart = [
        ds[var].quantile(0),
        ds[var].quantile(0.25),
        ds[var].quantile(0.50),
        ds[var].quantile(0.75),
        ds[var].quantile(1)
    ]

    summary = {
        "Average": mean,
        "Median": median,
        "Mode": mode,
        "Minimum": Quart[0],
        "Maximum": Quart[4],
        "Range": range,
        "Standard Deviation": std_dev,
        "Variance": variance,
        "Skewness": skew,
        "Kurtosis": kurtosis,
        "25th Percentile": Quart[1],
        "50th Percentile": Quart[2],
        "75th Percentile": Quart[3]
    }

    return summary
```

```python
# Regression: Loading and Selecting Relevant Columns

regression_df = df[[
    'study_hours',
    'attendance_percentage',
    'age',
    'math_score',
    'science_score',
    'english_score',
    'overall_score'
]].copy()

print("Regression Dataset:")
print(regression_df.head())

print("\nDataset Shape:", regression_df.shape)
print("\nMissing Values:")
print(regression_df.isnull().sum())
```

```python
# Handle Missing Values

print("Missing Values Before Treatment:")
print(regression_df.isnull().sum())

for column in regression_df.columns:
    regression_df[column] = regression_df[column].fillna(
        regression_df[column].mean()
    )

print("\nMissing Values After Treatment:")
print(regression_df.isnull().sum())
```

```python
# Detect Outliers Using the IQR Method

independent_columns = [
    'study_hours',
    'attendance_percentage',
    'age',
    'math_score',
    'science_score',
    'english_score'
]

for column in independent_columns:
    Q1 = regression_df[column].quantile(0.25)
    Q3 = regression_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = regression_df[
        (regression_df[column] < lower_bound) |
        (regression_df[column] > upper_bound)
    ]

    print("\nColumn:", column)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Number of Outliers:", len(outliers))
```

```python
# Combine Outliers from All Independent Variables

outlier_list = []

for column in independent_columns:
    Q1 = regression_df[column].quantile(0.25)
    Q3 = regression_df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = regression_df[
        (regression_df[column] < lower_bound) |
        (regression_df[column] > upper_bound)
    ]

    outlier_list.append(outliers)

all_outliers = pd.concat(outlier_list).drop_duplicates()

print("Total Unique Outlier Rows:", len(all_outliers))
```

```python
# Remove Outlier Rows

regression_df_clean = regression_df.drop(
    index=all_outliers.index
).copy()

print("Dataset Shape Before Removing Outliers:", regression_df.shape)
print("Dataset Shape After Removing Outliers:", regression_df_clean.shape)
print("Remaining Records:", len(regression_df_clean))
```

```python
# Splitting the Data

# Independent Variables (X)
X = regression_df_clean[[
    'study_hours',
    'attendance_percentage',
    'age',
    'math_score',
    'science_score',
    'english_score'
]]

# Target Variable (y)
y = regression_df_clean['overall_score']

# Display Independent Variables
print("Independent Variables (X):")
print(X.head())

# Display Target Variable
print("\nTarget Variable (y):")
print(y.head())

# Display Dataset Shapes
print("\nIndependent Variables Shape:", X.shape)
print("Target Variable Shape:", y.shape)
```

```python
# Train-Test Split (80% Training, 20% Testing)

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42
)

print("Training Features Shape:", X_train.shape)
print("Testing Features Shape:", X_test.shape)
print("Training Target Shape:", y_train.shape)
print("Testing Target Shape:", y_test.shape)
```

```python
# 1. Linear Regression

from sklearn.linear_model import LinearRegression

linear_model = LinearRegression()

# Train the model
linear_model.fit(X_train, y_train)

# Make predictions
y_pred_linear = linear_model.predict(X_test)

print("Linear Regression Model Trained Successfully")
print("Intercept:", linear_model.intercept_)
print("Coefficients:", linear_model.coef_)
```

```python
# 2. Polynomial Regression (Degree 2)

from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression

# Transform features into degree 2 polynomial features
poly = PolynomialFeatures(degree=2)

X_train_poly = poly.fit_transform(X_train)
X_test_poly = poly.transform(X_test)

# Train the Polynomial Regression model
poly_model = LinearRegression()
poly_model.fit(X_train_poly, y_train)

# Make predictions
y_pred_poly = poly_model.predict(X_test_poly)

print("Polynomial Regression (Degree 2) Trained Successfully")
print("Training Features Shape:", X_train_poly.shape)
print("Testing Features Shape:", X_test_poly.shape)
```

```python
# 3. Ridge Regression

from sklearn.linear_model import Ridge

# Create Ridge Regression model
ridge_model = Ridge(alpha=1.0)

# Train the model
ridge_model.fit(X_train, y_train)

# Make predictions
y_pred_ridge = ridge_model.predict(X_test)

print("Ridge Regression Model Trained Successfully")
print("Intercept:", ridge_model.intercept_)
print("Coefficients:", ridge_model.coef_)
```

```python
# 4. Lasso Regression

from sklearn.linear_model import Lasso

# Create Lasso Regression model
lasso_model = Lasso(alpha=1.0, max_iter=10000)

# Train the model
lasso_model.fit(X_train, y_train)

# Make predictions
y_pred_lasso = lasso_model.predict(X_test)

print("Lasso Regression Model Trained Successfully")
print("Intercept:", lasso_model.intercept_)
print("Coefficients:", lasso_model.coef_)
```

```python
# 5. Elastic Net Regression

from sklearn.linear_model import ElasticNet

# Create Elastic Net model
elastic_model = ElasticNet(alpha=1.0, l1_ratio=0.5, max_iter=10000)

# Train the model
elastic_model.fit(X_train, y_train)

# Make predictions
y_pred_elastic = elastic_model.predict(X_test)

print("Elastic Net Regression Model Trained Successfully")
print("Intercept:", elastic_model.intercept_)
print("Coefficients:", elastic_model.coef_)
```

```python
# Evaluating Models

from sklearn.metrics import mean_squared_error, r2_score

# Compare all five regression models
models = {
    "Linear Regression": y_pred_linear,
    "Polynomial Regression": y_pred_poly,
    "Ridge Regression": y_pred_ridge,
    "Lasso Regression": y_pred_lasso,
    "Elastic Net Regression": y_pred_elastic
}

# Calculate MSE and R-squared
for name, predictions in models.items():
    mse = mean_squared_error(y_test, predictions)
    r2 = r2_score(y_test, predictions)

    print("\n", name)
    print("Mean Squared Error (MSE):", mse)
    print("R-squared (R2):", r2)
```

```python
# Selecting the Best Model

import pandas as pd
from sklearn.metrics import mean_squared_error, r2_score

# Create comparison table
comparison_data = []

for name, predictions in models.items():
    mse = mean_squared_error(y_test, predictions)
    r2 = r2_score(y_test, predictions)

    comparison_data.append({
        "Model": name,
        "MSE": mse,
        "R2": r2
    })

comparison_df = pd.DataFrame(comparison_data)

# Display comparison table
print("Regression Models Comparison:")
print(comparison_df)

# Select the model with the highest R2
best_model = comparison_df.loc[comparison_df["R2"].idxmax()]

print("\nBest-Fit Model:", best_model["Model"])
print("Best R2 Score:", best_model["R2"])
```

```python
X = df[['study_hours', 'attendance_percentage', 'age', 'overall_score']].copy()

print(X.head())
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import pandas as pd

sc = StandardScaler()
X_scaled = sc.fit_transform(X)
```

```python
wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.plot(range(1, 11), wcss)
plt.title('The Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42)

y_kmeans = kmeans.fit_predict(X_scaled)

df['Cluster'] = y_kmeans

print(df['Cluster'].value_counts().sort_index())
```

```python
from sklearn.decomposition import PCA

pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_kmeans, cmap='rainbow')

plt.title('Student Performance Clusters')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.show()
```

```python
cluster_analysis = df.groupby('Cluster')[
    ['study_hours', 'attendance_percentage', 'age', 'overall_score']
].mean()

print(cluster_analysis)
```
